```python
from flask import Flask, render_template
import pandas as pd
import os

app = Flask(__name__)

# Load dataset
df = pd.read_csv("All Time Men ODI Most Hundreds Career 2025.csv")

# Convert numeric columns
columns = [
    "Total_Runs",
    "Matches_Played",
    "Centuries_100s",
    "Batting_Average",
    "Strike_Rate"
]

for col in columns:
    df[col] = pd.to_numeric(df[col], errors="coerce").fillna(0)


# Player image folder
image_folder = "static/images/players"

# Player names and image filenames
player_images = {
    "AB de Villiers": "Ab de Villiers",
    "DPMD Jayawardene": "DPMD Jayawardene",
    "HM Amla": "HM Amla",
    "Inzamam-ul-Haq": "inzamam ul haq",
    "KC Sangakkara": "KC Sangakkara",
    "MS Dhoni": "MS Dhoni",
    "Q de Kock": "Q de Kock",
    "R Dravid": "R Dravid",
    "RG Sharma": "rohit sharma",
    "RT Ponting": "RT Ponting",
    "SR Tendulkar": "SR Tendulkar",
    "ST Jayasuriya": "St Jayasuriya",
    "V Kohli": "V Kohli",
    "JH Kallis": "JH Kallis",
    "SC Ganguly": "SC Ganguly"
}


# Find player image
def get_player_image(player):

    image_name = player_images.get(player)

    if not image_name or not os.path.isdir(image_folder):
        return None

    for file in os.listdir(image_folder):

        filename = os.path.splitext(file)[0]

        if filename.strip().lower() == image_name.strip().lower():
            return file

    return None


# ODI World Cup winners
world_cup_winners = [
    {"Year": 1975, "Winner": "West Indies", "Runner_Up": "Australia", "Captain": "Clive Lloyd"},
    {"Year": 1979, "Winner": "West Indies", "Runner_Up": "England", "Captain": "Clive Lloyd"},
    {"Year": 1983, "Winner": "India", "Runner_Up": "West Indies", "Captain": "Kapil Dev"},
    {"Year": 1987, "Winner": "Australia", "Runner_Up": "England", "Captain": "Allan Border"},
    {"Year": 1992, "Winner": "Pakistan", "Runner_Up": "England", "Captain": "Imran Khan"},
    {"Year": 1996, "Winner": "Sri Lanka", "Runner_Up": "Australia", "Captain": "Arjuna Ranatunga"},
    {"Year": 1999, "Winner": "Australia", "Runner_Up": "Pakistan", "Captain": "Steve Waugh"},
    {"Year": 2003, "Winner": "Australia", "Runner_Up": "India", "Captain": "Ricky Ponting"},
    {"Year": 2007, "Winner": "Australia", "Runner_Up": "Sri Lanka", "Captain": "Ricky Ponting"},
    {"Year": 2011, "Winner": "India", "Runner_Up": "Sri Lanka", "Captain": "MS Dhoni"},
    {"Year": 2015, "Winner": "Australia", "Runner_Up": "New Zealand", "Captain": "Michael Clarke"},
    {"Year": 2019, "Winner": "England", "Runner_Up": "New Zealand", "Captain": "Eoin Morgan"},
    {"Year": 2023, "Winner": "Australia", "Runner_Up": "India", "Captain": "Pat Cummins"}
]


@app.route("/")
def home():

    # Dashboard KPIs
    players = len(df)
    runs = int(df["Total_Runs"].sum())
    centuries = int(df["Centuries_100s"].sum())
    matches = int(df["Matches_Played"].sum())

    average_runs = round(df["Total_Runs"].mean(), 2)
    average_strike_rate = round(df["Strike_Rate"].mean(), 2)

    # Record holders
    highest_run_scorer = df.loc[df["Total_Runs"].idxmax()].to_dict()
    highest_average = df.loc[df["Batting_Average"].idxmax()].to_dict()
    highest_strike_rate = df.loc[df["Strike_Rate"].idxmax()].to_dict()
    most_centuries = df.loc[df["Centuries_100s"].idxmax()].to_dict()

    # All player profile cards
    all_players = df.sort_values(
        "Total_Runs", ascending=False
    ).to_dict("records")

    for player in all_players:
        player["image"] = get_player_image(player["Player"])

    # Country filter
    countries = sorted(df["Country"].dropna().unique().tolist())

    # Country-wise player count and runs
    country_data = (
        df.groupby("Country")
        .agg(
            Players=("Player", "count"),
            Total_Runs=("Total_Runs", "sum"),
            Total_Centuries=("Centuries_100s", "sum")
        )
        .reset_index()
        .sort_values("Total_Runs", ascending=False)
        .to_dict("records")
    )

    # World Cup title count
    title_counts = {}

    for item in world_cup_winners:
        winner = item["Winner"]
        title_counts[winner] = title_counts.get(winner, 0) + 1

    team_titles = [
        {"Team": team, "Titles": count}
        for team, count in sorted(
            title_counts.items(),
            key=lambda x: x[1],
            reverse=True
        )
    ]

    return render_template(
        "index.html",
        players=players,
        runs=runs,
        centuries=centuries,
        matches=matches,
        average_runs=average_runs,
        average_strike_rate=average_strike_rate,
        highest_run_scorer=highest_run_scorer,
        highest_average=highest_average,
        highest_strike_rate=highest_strike_rate,
        most_centuries=most_centuries,
        all_players=all_players,
        countries=countries,
        country_data=country_data,
        world_cup_winners=world_cup_winners,
        team_titles=team_titles
    )


if __name__ == "__main__":
    app.run(
        host="127.0.0.1",
        port=5000,
        debug=False,
        use_reloader=False
    )
```
